```python
# Install OpenCV if not already installed
!pip install opencv-python

# Import libraries
import cv2
from google.colab.patches import cv2_imshow

# Load pre-trained face detection model (Haar Cascade)
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Load an image from local files
img = cv2.imread('/content/istockphoto-1368965646-612x612.jpg')

# Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces in the image
faces = face_cascade.detectMultiScale(gray, 1.3, 5)

# Draw rectangles around detected faces
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)

# Show the final image with detected faces
cv2_imshow(img)

# Wait for a key press and close windows (not essential in Colab)
cv2.waitKey(0)
cv2.destroyAllWindows()

#PS: Haar is not rotation-invariant. So it did not recognize some slightly tilted faces.
```
